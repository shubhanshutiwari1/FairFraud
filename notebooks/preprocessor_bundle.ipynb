```python
import joblib
import os
import sys

# 1. Force the absolute path to your Movies folder project
PROJECT_ROOT = r"C:\Users\shubh\projects\FairFraud"
MODELS_DIR = os.path.join(PROJECT_ROOT, "models")
os.makedirs(MODELS_DIR, exist_ok=True)

if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

from src.pipeline import get_data
data = get_data()

# 2. Extract only the fitted preprocessor objects
minimal_bundle = {
    "scaler": data["scaler"],
    "num_imputer": data["num_imputer"],
    "cat_imputer": data["cat_imputer"],
    "onehot_encoder": data["onehot_encoder"],
    "categorical_features": data.get("categorical_features", [])
}

# 3. Save directly to the exact models folder
output_path = os.path.join(MODELS_DIR, "preprocessor_bundle.pkl")
joblib.dump(minimal_bundle, output_path)

print(f"Bundle successfully saved to: {output_path}")
print(f"File size: {os.path.getsize(output_path) / 1024:.2f} KB")
```
